```python
# Step 1 - Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2 - Load the dataset
df = pd.read_csv("india-districts-census-2011.csv")

# Step 3 - See the data
print(df.head())        # first 5 rows
print(df.shape)         # rows and columns count
print(df.columns)       # column names
```

```python
# See basic info about dataset
print("Shape of dataset:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())
print("\nBasic Statistics:")
print(df.describe())
```

```python
# Check missing values
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Top 10 populated states
plt.figure(figsize=(12,6))
top10 = df.groupby('State')['Population'].sum().sort_values(ascending=False).head(10)
top10.plot(kind='bar', color='steelblue')
plt.title('Top 10 Populated States in India', fontsize=16)
plt.xlabel('State', fontsize=12)
plt.ylabel('Population', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
top10 = df.groupby('State name')['Population'].sum().sort_values(ascending=False).head(10)
top10.plot(kind='bar', color='steelblue')
plt.title('Top 10 Populated States in India', fontsize=16)
plt.xlabel('State', fontsize=12)
plt.ylabel('Population', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
gender_data = [df['Male'].sum(), df['Female'].sum()]
labels = ['Male', 'Female']
colors = ['steelblue', 'pink']

plt.figure(figsize=(7,7))
plt.pie(gender_data, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Male vs Female Population in India', fontsize=16)
plt.show()
```

```python
plt.figure(figsize=(12,6))
literacy = df.groupby('State name')['Literate'].sum().sort_values(ascending=False).head(10)
literacy.plot(kind='barh', color='green')
plt.title('Top 10 States by Literacy in India', fontsize=16)
plt.xlabel('Literate Population', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))
male_lit = df.groupby('State name')['Male_Literate'].sum().sort_values(ascending=False).head(10)
female_lit = df.groupby('State name')['Female_Literate'].sum().sort_values(ascending=False).head(10)

x = range(len(male_lit))
plt.bar(x, male_lit.values, width=0.4, label='Male Literate', color='steelblue')
plt.bar([i+0.4 for i in x], female_lit.values, width=0.4, label='Female Literate', color='pink')
plt.xticks([i+0.2 for i in x], male_lit.index, rotation=45)
plt.title('Male vs Female Literacy by State', fontsize=16)
plt.legend()
plt.tight_layout()
plt.show()
```
